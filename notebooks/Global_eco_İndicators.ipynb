# **Hypotesis Testing**

# 1. Household consumption ilə Government consumption arasında müqayisə **(Paired (asılı) t-test)**

**İqtisadiyyatda əsas xərclər daha çox ev təsərrüfatları tərəfindən edilir, yoxsa dövlət tərəfindən?**

```python
from scipy import stats

household = df['household_consumption']
government = df['government_consumption']

t_stat, p_value = stats.ttest_rel(household, government)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

t-stat:ferqin ne qeder boyuk oldugunu olcur

p_value:eger arada hec bir ferq yoxdursa,tesadufen gorme ehtimalimiz ne qederdir

**H₀ (Null hipotez):**  
Household consumption ilə government consumption arasında orta baxımından fərq yoxdur.

**H₁ (Alternativ hipotez):**  
Household consumption ilə government consumption arasında orta baxımından fərq vardır.

p-value 0.05-dən çox kiçik olduğu üçün **null hipotez rədd edilir**.  
Bu nəticə household consumption ilə government consumption arasında **statistik olaraq güclü fərqin** olduğunu göstərir.


# **2**.GDP və GNI Arasında **Pearson Correlation Analysis**

ÜDM (GDP) ilə Milli Gəlir (GNI) arasında statistik əlaqənin mövcudurmu?

```python
from scipy import stats

gdp = df['gdp']
gni = df['gni']

r_value, p_value = stats.pearsonr(gdp, gni)

print("correlation coefficient (r):", r_value)
print("p-value:", p_value)
```

r_value-nin 1-ə çox yaxın olması GDP ilə GNI arasında müsbət əlaqənin mövcud olduğunu göstərir.

 p-value-nun çox kiçik olması bu əlaqənin statistik olaraq əhəmiyyətli olduğunu göstərir.

 **GDP ilə GNI arasında çox güclü və gözlənilən struktur əlaqə mövcuddur.**

# 3. Population və GDP Per Capita arasında **Pearson Correlation Analysis**

Əhali artımı ilə iqtisadi rifah arasında əlaqə mövcuddurmu?

```python
from scipy import stats

population = df['population']
gdp_pc = df['gdp_per_capita']

r_value, p_value = stats.pearsonr(population, gdp_pc)

print("correlation coefficient (r):", r_value)
print("p-value:", p_value)
```

r < 0.1

Korrelyasiya əmsalı sıfıra çox yaxın olduğu üçün bu iki göstərici arasında güclü əlaqə olmadığı görülür. Əgər r yüksək olsaydı, bu, əhali artdıqca rifah səviyyəsinin də artdığını göstərəcəkdi. Lakin nəticə göstərir ki, əhali ölçüsü iqtisadi rifahın artımında rol oynamır
